### -> IMPORT LIBRARIE

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.preprocessing import StandardScaler, MinMaxScaler
import warnings
warnings.filterwarnings('ignore')

pd.set_option('display.max_columns', None)
plt.style.use('seaborn-v0_8')
sns.set_palette('husl')
```

### -> LOAD DATASET


```python
df = pd.read_csv('../Data/cardio_train.csv', delimiter=';')

print("Dataset Shape:", df.shape)
print("\nFirst 5 rows of the dataset:")
df.head()
```

### -> HANDLE MISSING VALUES

```python
print("Missing Values Count:")
print(df.isnull().sum())
print("\nTotal Missing Values:", df.isnull().sum().sum())
```

```python
print("Percentage of Missing Values:")
print((df.isnull().sum() / len(df)) * 100)
```

### -> CHECK AND REMOVE DUPLICATES

```python
#DROP ID COLUMN
print("Checking for ID column...")
print("="*60)

if 'id' in df.columns:
    df = df.drop('id', axis=1)
    print("ID column removed")
    print(f"\nNew Dataset Shape: {df.shape}")
else:
    print("No 'id' column found in dataset")
```

```python
#CHECK AND REMOVE DUPLICATES
print("Checking for duplicate rows...")
print("="*60)

# Check duplicates
duplicates = df.duplicated().sum()
print(f"\nTotal Duplicate Rows: {duplicates}")
print(f"Percentage: {(duplicates/len(df))*100:.2f}%")

# Show duplicate rows if any
if duplicates > 0:
    print(f"\nSample of duplicate rows:")
    print(df[df.duplicated(keep=False)].head(10))
```

```python
# Remove duplicates
print("Before Removing Duplicates:", df.shape)

df = df.drop_duplicates()

print("After Removing Duplicates:", df.shape)
print(f"Removed: {duplicates} duplicate rows")
print("Duplicate removal completed!")
```

### -> IDENTIFY AND HANDLE OUTLIERS

```python
def detect_outliers_iqr(data, column):
    Q1 = data[column].quantile(0.25)
    Q3 = data[column].quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    outliers = data[(data[column] < lower_bound) | (data[column] > upper_bound)]
    return outliers, lower_bound, upper_bound
```

```python
numerical_cols = ['age', 'height', 'weight', 'ap_hi', 'ap_lo']

print("Outlier Detection Summary (IQR Method):")
print("="*60)
for col in numerical_cols:
    outliers, lower, upper = detect_outliers_iqr(df, col)
    print(f"\n{col}:")
    print(f"  Lower Bound: {lower:.2f}, Upper Bound: {upper:.2f}")
    print(f"  Outliers: {len(outliers)} ({(len(outliers)/len(df))*100:.2f}%)")
```

```python
fig, axes = plt.subplots(2, 3, figsize=(15, 10))
fig.suptitle('Outlier Detection - Boxplots', fontsize=14, fontweight='bold')

for idx, col in enumerate(numerical_cols):
    row = idx // 3
    col_idx = idx % 3
    axes[row, col_idx].boxplot(df[col].dropna())
    axes[row, col_idx].set_title(f'{col}', fontweight='bold')
    axes[row, col_idx].set_ylabel('Value')
    axes[row, col_idx].grid(True, alpha=0.3)

fig.delaxes(axes[1, 2])
plt.tight_layout()
plt.show()
```

```python
print("Original Dataset Shape:", df.shape)

df_cleaned = df.copy()

# Remove invalid blood pressure
df_cleaned = df_cleaned[df_cleaned['ap_hi'] > df_cleaned['ap_lo']]

# Remove unrealistic blood pressure values
df_cleaned = df_cleaned[(df_cleaned['ap_hi'] >= 80) & (df_cleaned['ap_hi'] <= 200)]
df_cleaned = df_cleaned[(df_cleaned['ap_lo'] >= 50) & (df_cleaned['ap_lo'] <= 140)]

# Remove unrealistic height and weight
df_cleaned = df_cleaned[(df_cleaned['height'] >= 140) & (df_cleaned['height'] <= 220)]
df_cleaned = df_cleaned[(df_cleaned['weight'] >= 30) & (df_cleaned['weight'] <= 200)]

print("Cleaned Dataset Shape:", df_cleaned.shape)
print(f"Removed: {len(df) - len(df_cleaned)} records ({((len(df) - len(df_cleaned))/len(df))*100:.2f}%)")
```

### -> ENCODE CATEGORICAL VARIABLES

```python
categorical_cols = ['gender', 'cholesterol', 'gluc', 'smoke', 'alco', 'active', 'cardio']

print("Categorical Variables:")
for col in categorical_cols:
    print(f"\n{col}:")
    print(df_cleaned[col].value_counts())
```

```python
print("Encoding Information:")
print("\ngender: 1 = female, 2 = male")
print("cholesterol: 1 = normal, 2 = above normal, 3 = well above normal")
print("gluc: 1 = normal, 2 = above normal, 3 = well above normal")
print("smoke: 0 = no, 1 = yes")
print("alco: 0 = no, 1 = yes")
print("active: 0 = no, 1 = yes")
print("cardio: 0 = no disease, 1 = disease")
```

```python
# Convert gender to 0 and 1
df_cleaned['gender'] = df_cleaned['gender'] - 1
print("\nGender recoded: 0 = female, 1 = male")
```

### -> FEATURE ENGINEERING

```python
# Convert age from days to years
df_cleaned['age_years'] = (df_cleaned['age'] / 365).round().astype(int)

# Calculate BMI
df_cleaned['bmi'] = df_cleaned['weight'] / ((df_cleaned['height'] / 100) ** 2)

# Calculate Pulse Pressure
df_cleaned['pulse_pressure'] = df_cleaned['ap_hi'] - df_cleaned['ap_lo']

# Calculate Mean Arterial Pressure
df_cleaned['map'] = df_cleaned['ap_lo'] + (df_cleaned['pulse_pressure'] / 3)

print("New Features Created:")
print("- age_years: Age in years")
print("- bmi: Body Mass Index")
print("- pulse_pressure: Pulse Pressure")
print("- map: Mean Arterial Pressure")
```

```python
print("\nSample of new features:")
df_cleaned[['age_years', 'bmi', 'pulse_pressure', 'map']].head()
```

### -> NORMALIZE/SCALE NUMERICAL FEATURES

```python
features_to_scale = ['age', 'height', 'weight', 'ap_hi', 'ap_lo', 'bmi', 'pulse_pressure', 'map']

print("Statistics Before Scaling:")
df_cleaned[features_to_scale].describe()
```

```python
# StandardScaler
df_scaled = df_cleaned.copy()
scaler_standard = StandardScaler()
df_scaled[features_to_scale] = scaler_standard.fit_transform(df_cleaned[features_to_scale])

print("Statistics After Standard Scaling:")
df_scaled[features_to_scale].describe()
```

```python
# MinMaxScaler
df_minmax = df_cleaned.copy()
scaler_minmax = MinMaxScaler()
df_minmax[features_to_scale] = scaler_minmax.fit_transform(df_cleaned[features_to_scale])

print("Statistics After MinMax Scaling:")
df_minmax[features_to_scale].describe()
```

```python
fig, axes = plt.subplots(1, 3, figsize=(18, 5))
fig.suptitle('Comparison of Scaling Methods', fontsize=14, fontweight='bold')

df_cleaned[features_to_scale[:4]].boxplot(ax=axes[0])
axes[0].set_title('Original Data', fontweight='bold')
axes[0].tick_params(axis='x', rotation=45)

df_scaled[features_to_scale[:4]].boxplot(ax=axes[1])
axes[1].set_title('Standard Scaler', fontweight='bold')
axes[1].tick_params(axis='x', rotation=45)

df_minmax[features_to_scale[:4]].boxplot(ax=axes[2])
axes[2].set_title('MinMax Scaler', fontweight='bold')
axes[2].tick_params(axis='x', rotation=45)

plt.tight_layout()
plt.show()
```

```python
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

df_cleaned[df_cleaned['cardio']==0]['bmi'].hist(ax=axes[0], bins=30, alpha=0.6, label='No Disease', color='lightblue')
df_cleaned[df_cleaned['cardio']==1]['bmi'].hist(ax=axes[0], bins=30, alpha=0.6, label='Disease', color='salmon')
axes[0].set_title('BMI Distribution by Cardiovascular Disease', fontweight='bold')
axes[0].set_xlabel('BMI')
axes[0].set_ylabel('Frequency')
axes[0].legend()
axes[0].grid(True, alpha=0.3)

df_cleaned.boxplot(column='bmi', by='cardio', ax=axes[1])
axes[1].set_title('BMI by Cardiovascular Disease', fontweight='bold')
axes[1].set_xlabel('Cardio (0=No, 1=Yes)')
axes[1].set_ylabel('BMI')
plt.suptitle('')

plt.tight_layout()
plt.show()
```

```python
categorical_features = ['gender', 'cholesterol', 'gluc', 'smoke', 'alco', 'active']

fig, axes = plt.subplots(2, 3, figsize=(16, 10))
fig.suptitle('Categorical Features vs Cardiovascular Disease', fontsize=14, fontweight='bold')

for idx, col in enumerate(categorical_features):
    row = idx // 3
    col_idx = idx % 3
    
    pd.crosstab(df_cleaned[col], df_cleaned['cardio'], normalize='index').plot(
        kind='bar', ax=axes[row, col_idx], stacked=False, color=['lightblue', 'salmon'])
    axes[row, col_idx].set_title(f'{col}', fontweight='bold')
    axes[row, col_idx].set_xlabel('')
    axes[row, col_idx].set_ylabel('Proportion')
    axes[row, col_idx].legend(['No Disease', 'Disease'])
    axes[row, col_idx].tick_params(axis='x', rotation=0)

plt.tight_layout()
plt.show()
```
