### -> IMPORT NECESSARY LIBRARIES

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Set visualization style
plt.style.use('seaborn-v0_8-darkgrid')
sns.set_palette("husl")
```

### -> LOAD THE DATASET

```python
df = pd.read_csv('../Data/cardio_train.csv', sep=';')
print(f"Dataset shape: {df.shape[0]} rows × {df.shape[1]} columns")
```

### -> DATASET STRUCTURE & INFORMATION

```python
print("\nColumn Information:")
df.info()
```

```python
print("\nExtended Statistics (including percentiles):")
df.describe(percentiles=[.01, .05, .25, .5, .75, .95, .99])
```

### -> Feature Understanding

```python
feature_descriptions = {
    'id': 'Unique patient identifier',
    'age': 'Age in days (need to convert to years)',
    'gender': 'Gender (1: Female, 2: Male)',
    'height': 'Height in centimeters',
    'weight': 'Weight in kilograms',
    'ap_hi': 'Systolic blood pressure',
    'ap_lo': 'Diastolic blood pressure',
    'cholesterol': 'Cholesterol level (1: Normal, 2: Above normal, 3: Well above normal)',
    'gluc': 'Glucose level (1: Normal, 2: Above normal, 3: Well above normal)',
    'smoke': 'Smoking status (0: No, 1: Yes)',
    'alco': 'Alcohol intake (0: No, 1: Yes)',
    'active': 'Physical activity (0: No, 1: Yes)',
    'cardio': 'Target variable - Cardiovascular disease (0: No, 1: Yes)'
}

for col, desc in feature_descriptions.items():
    print(f"  • {col:15s}: {desc}")
```

### -> Missing Values Analysis

```python
missing_values = df.isnull().sum()
missing_percentage = (df.isnull().sum() / len(df)) * 100

missing_df = pd.DataFrame({
    'Column': df.columns,
    'Missing_Count': missing_values.values,
    'Missing_Percentage': missing_percentage.values
})

print("\nMissing Values Report:")
print(missing_df)
```

### -> Duplicate Values Analysis

```python
duplicate_count = df.duplicated().sum()
duplicate_percentage = (duplicate_count / len(df)) * 100

print(f"\nTotal Duplicate Rows: {duplicate_count:,}")
print(f"Percentage of Duplicates: {duplicate_percentage:.2f}%")

if duplicate_count > 0:
    print(f"\nWARNING: Found {duplicate_count} duplicate rows!")
else:
    print("\nGOOD NEWS: No duplicate rows found in the dataset!")
```

### -> Target Variable Distribution

```python
print("\nCardiovascular Disease Distribution:")
cardio_counts = df['cardio'].value_counts()
cardio_percentage = df['cardio'].value_counts(normalize=True) * 100

target_summary = pd.DataFrame({
    'Disease_Status': ['No Disease (0)', 'Disease Present (1)'],
    'Count': cardio_counts.values,
    'Percentage': cardio_percentage.values
})

print(target_summary)

# Calculate balance ratio
balance_ratio = cardio_counts.min() / cardio_counts.max()
print(f"\nClass Balance Ratio: {balance_ratio:.2f}")

if balance_ratio > 0.8:
    print("Dataset is WELL BALANCED")
elif balance_ratio > 0.5:
    print("Dataset is MODERATELY BALANCED")
else:
    print("Dataset is IMBALANCED")
```

### -> Numerical Features Analysis

```python
numerical_cols = df.select_dtypes(include=[np.number]).columns.tolist()
print(f"\nNumerical Features ({len(numerical_cols)}):")
print(numerical_cols)

print("\nValue Ranges:")
for col in numerical_cols:
    print(f"  • {col:15s}: Min = {df[col].min():10.2f}, Max = {df[col].max():10.2f}")
```

### -> Categorical Features Analysis

```python
categorical_cols = ['gender', 'cholesterol', 'gluc', 'smoke', 'alco', 'active']

print(f"\nCategorical Features ({len(categorical_cols)}):")
for col in categorical_cols:
    print(f"\n  • {col.upper()}:")
    print(df[col].value_counts().to_string())
    print(f"    Unique values: {df[col].nunique()}")
```

### -> Initial Observations & Key Insights

```python
observations = f"""
KEY FINDINGS:

1. DATASET CHARACTERISTICS:
   • Total records: {df.shape[0]:,} patients
   • No missing values detected
   • Duplicate rows: {df.duplicated().sum():,}
   • Target variable is well-balanced (approximately 50-50 split)

2. DATA QUALITY ISSUES (to be addressed in Section 2):
   • Age is in DAYS - needs conversion to YEARS
   • Possible outliers in:
     - Blood pressure (negative or extremely high values)
     - Height and weight (unrealistic values)
   • Need to calculate BMI from height and weight
   • Remove duplicate records if any

3. FEATURE ENGINEERING OPPORTUNITIES:
   • Create BMI = weight / (height/100)²
   • Convert age from days to years
   • Create age groups (Young, Middle-aged, Senior)
   • Create BP categories (Normal, Elevated, High)

4. PRELIMINARY ANALYSIS NEEDED:
   • Correlation analysis between features
   • Distribution analysis of numerical features
   • Outlier detection and treatment strategy
   • Feature importance analysis

5. MODELING CONSIDERATIONS:
   • Binary classification problem
   • Balanced dataset - no need for resampling initially
   • Mix of numerical and categorical features
   • Consider algorithms: Logistic Regression, Random Forest, XGBoost, Neural Networks
"""

print(observations)
```

### -> Visualizations

```python
print("\n" + "=" * 80)
print("CREATING INITIAL VISUALIZATIONS")
print("=" * 80)

# ============================================================================
# PART 1: ORIGINAL 4 VISUALIZATIONS (Dashboard Style)

# Create figure with 4 subplots
fig, axes = plt.subplots(2, 2, figsize=(15, 10))
fig.suptitle('Dataset Overview Dashboard', fontsize=16, fontweight='bold')

# --- GRAPH 1: Target Variable (Bar Chart) ---
cardio_counts = df['cardio'].value_counts()
axes[0, 0].bar([0, 1], cardio_counts.values, color=['#2ecc71', '#e74c3c'], width=0.6)
axes[0, 0].set_title('Disease Distribution', fontweight='bold')
axes[0, 0].set_xlabel('Disease Status')
axes[0, 0].set_ylabel('Count')
axes[0, 0].set_xticks([0, 1])
axes[0, 0].set_xticklabels(['No Disease', 'Disease'])

# Add numbers on bars
for i, v in enumerate(cardio_counts.values):
    axes[0, 0].text(i, v + 500, f'{v:,}\n({v/len(df)*100:.1f}%)', 
                    ha='center', fontweight='bold')

# --- GRAPH 2: Gender Distribution (Pie Chart) ---
gender_counts = df['gender'].value_counts()
axes[0, 1].pie(gender_counts.values, 
               labels=['Female', 'Male'], 
               autopct='%1.1f%%',
               colors=['#ff69b4', '#4169e1'], 
               startangle=90)
axes[0, 1].set_title('Gender Distribution', fontweight='bold')

# --- GRAPH 3: Feature Types (Bar Chart) ---
num_features = len(df.select_dtypes(include=[np.number]).columns) - 1  # exclude 'id'
cat_features = 6  # gender, cholesterol, gluc, smoke, alco, active

axes[1, 0].bar(['Numerical', 'Categorical'], [num_features, cat_features], 
               color=['#3498db', '#e67e22'], width=0.5)
axes[1, 0].set_title('Feature Types', fontweight='bold')
axes[1, 0].set_ylabel('Count')

# Add numbers on bars
axes[1, 0].text(0, num_features + 0.2, str(num_features), ha='center', fontweight='bold')
axes[1, 0].text(1, cat_features + 0.2, str(cat_features), ha='center', fontweight='bold')

# --- GRAPH 4: Data Quality (Bar Chart) ---
complete_records = len(df) - df.isnull().any(axis=1).sum()
missing_values = df.isnull().any(axis=1).sum()
duplicates = df.duplicated().sum()

quality_data = [complete_records, missing_values, duplicates]
quality_labels = ['Complete\nRecords', 'Missing\nValues', 'Duplicate\nRecords']

axes[1, 1].bar(quality_labels, quality_data, 
               color=['#2ecc71', '#e74c3c', '#f39c12'], width=0.5)
axes[1, 1].set_title('Data Quality', fontweight='bold')
axes[1, 1].set_ylabel('Count')

# Add numbers on bars
for i, v in enumerate(quality_data):
    axes[1, 1].text(i, v + 1000, f'{v:,}', ha='center', fontweight='bold')

plt.tight_layout()
plt.savefig('dashboard_4graphs.png', dpi=300, bbox_inches='tight')
plt.show()

print("✓ Saved: dashboard_4graphs.png")
```

```python
# ============================================================================
# PART 2: SIMPLE ADDITIONAL VISUALIZATIONS
# ============================================================================

# --- GRAPH 5: Age Distribution (Histogram) ---
print("\nCreating Age Distribution...")

plt.figure(figsize=(10, 6))

# Convert age from days to years
age_years = df['age'] / 365.25

# Plot histogram
plt.hist(age_years, bins=30, color='steelblue', edgecolor='black', alpha=0.7)
plt.xlabel('Age (years)', fontsize=12)
plt.ylabel('Number of Patients', fontsize=12)
plt.title('Age Distribution of Patients', fontsize=14, fontweight='bold')
plt.grid(alpha=0.3)

# Add mean line
mean_age = age_years.mean()
plt.axvline(mean_age, color='red', linestyle='--', linewidth=2, 
            label=f'Mean Age: {mean_age:.1f} years')
plt.legend()

plt.tight_layout()
plt.savefig('graph5_age_distribution.png', dpi=300, bbox_inches='tight')
plt.show()

print("Saved: graph5_age_distribution.png")

# --- GRAPH 6: Disease by Gender (Simple Bar Chart) ---
print("\nCreating Disease by Gender...")

plt.figure(figsize=(10, 6))

# Calculate percentages
female_data = df[df['gender'] == 1]
male_data = df[df['gender'] == 2]

female_disease = (female_data['cardio'].sum() / len(female_data)) * 100
male_disease = (male_data['cardio'].sum() / len(male_data)) * 100

# Plot
genders = ['Female', 'Male']
disease_rates = [female_disease, male_disease]

bars = plt.bar(genders, disease_rates, color=['#ff69b4', '#4169e1'], width=0.5)
plt.ylabel('Disease Rate (%)', fontsize=12)
plt.title('Cardiovascular Disease Rate by Gender', fontsize=14, fontweight='bold')
plt.ylim(0, 100)
plt.grid(axis='y', alpha=0.3)

# Add percentages on bars
for i, bar in enumerate(bars):
    height = bar.get_height()
    plt.text(bar.get_x() + bar.get_width()/2., height + 1,
             f'{disease_rates[i]:.1f}%', ha='center', fontweight='bold', fontsize=12)

plt.tight_layout()
plt.savefig('graph6_disease_by_gender.png', dpi=300, bbox_inches='tight')
plt.show()

print("Saved: graph6_disease_by_gender.png")

# --- GRAPH 7: Blood Pressure Comparison (Box Plot) ---
print("\nCreating Blood Pressure Comparison...")

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 6))
fig.suptitle('Blood Pressure - Disease vs No Disease', fontsize=14, fontweight='bold')

# Systolic BP
no_disease_sys = df[df['cardio'] == 0]['ap_hi']
disease_sys = df[df['cardio'] == 1]['ap_hi']

ax1.boxplot([no_disease_sys, disease_sys], 
            labels=['No Disease', 'Disease'],
            patch_artist=True,
            boxprops=dict(facecolor='lightblue'))
ax1.set_ylabel('Systolic BP', fontsize=12)
ax1.set_title('Systolic Blood Pressure', fontweight='bold')
ax1.grid(alpha=0.3)

# Diastolic BP
no_disease_dia = df[df['cardio'] == 0]['ap_lo']
disease_dia = df[df['cardio'] == 1]['ap_lo']

ax2.boxplot([no_disease_dia, disease_dia], 
            labels=['No Disease', 'Disease'],
            patch_artist=True,
            boxprops=dict(facecolor='lightcoral'))
ax2.set_ylabel('Diastolic BP', fontsize=12)
ax2.set_title('Diastolic Blood Pressure', fontweight='bold')
ax2.grid(alpha=0.3)

plt.tight_layout()
plt.savefig('graph7_blood_pressure.png', dpi=300, bbox_inches='tight')
plt.show()

print("Saved: graph7_blood_pressure.png")

# --- GRAPH 8: Cholesterol Levels (Stacked Bar) ---
print("\nCreating Cholesterol Analysis...")

plt.figure(figsize=(10, 6))

# Calculate data
chol_disease = pd.crosstab(df['cholesterol'], df['cardio'], normalize='index') * 100

# Plot
chol_disease.plot(kind='bar', stacked=False, color=['#2ecc71', '#e74c3c'], width=0.7)
plt.xlabel('Cholesterol Level', fontsize=12)
plt.ylabel('Percentage (%)', fontsize=12)
plt.title('Disease Rate by Cholesterol Level', fontsize=14, fontweight='bold')
plt.xticks([0, 1, 2], ['Normal', 'Above Normal', 'High'], rotation=0)
plt.legend(['No Disease', 'Disease'], title='Status')
plt.grid(axis='y', alpha=0.3)

plt.tight_layout()
plt.savefig('graph8_cholesterol.png', dpi=300, bbox_inches='tight')
plt.show()

print("Saved: graph8_cholesterol.png")

# --- GRAPH 9: Correlation Heatmap (Simple) ---
print("\nCreating Correlation Heatmap...")

plt.figure(figsize=(10, 8))

# Calculate correlation
correlation = df.corr()

# Plot heatmap
sns.heatmap(correlation, 
            annot=True,      # Show numbers
            fmt='.2f',       # 2 decimal places
            cmap='coolwarm', # Color scheme
            center=0,        # Center at 0
            square=True,     # Square cells
            linewidths=0.5)  # Lines between cells

plt.title('Correlation Heatmap - All Features', fontsize=14, fontweight='bold', pad=20)
plt.tight_layout()
plt.savefig('graph9_correlation.png', dpi=300, bbox_inches='tight')
plt.show()

print("Saved: graph9_correlation.png")

# --- GRAPH 10: Lifestyle Factors (Simple Grid) ---
print("\nCreating Lifestyle Factors Analysis...")

fig, axes = plt.subplots(1, 3, figsize=(15, 5))
fig.suptitle('Lifestyle Factors vs Disease', fontsize=14, fontweight='bold')

lifestyle_features = ['smoke', 'alco', 'active']
titles = ['Smoking', 'Alcohol', 'Physical Activity']

for i, (feature, title) in enumerate(zip(lifestyle_features, titles)):
    # Calculate percentages
    data = pd.crosstab(df[feature], df['cardio'], normalize='index') * 100
    
    # Plot
    data.plot(kind='bar', ax=axes[i], color=['#2ecc71', '#e74c3c'], width=0.6)
    axes[i].set_title(title, fontweight='bold')
    axes[i].set_xlabel('')
    axes[i].set_ylabel('Percentage (%)')
    axes[i].set_xticklabels(['No', 'Yes'], rotation=0)
    axes[i].legend(['No Disease', 'Disease'], fontsize=8)
    axes[i].grid(axis='y', alpha=0.3)

plt.tight_layout()
plt.savefig('graph10_lifestyle.png', dpi=300, bbox_inches='tight')
plt.show()

print("Saved: graph10_lifestyle.png")

# ============================================================================
# SUMMARY OF ALL VISUALIZATIONS
# ============================================================================

# Print key insights
print("\n" + "="*80)
print("KEY INSIGHTS FROM VISUALIZATIONS")
print("="*80)

print(f"""
1. Target Balance: Dataset is well-balanced ({cardio_counts[0]} vs {cardio_counts[1]})
2. Gender: {len(female_data)} females, {len(male_data)} males
3. Average Age: {age_years.mean():.1f} years
4. Disease Rate - Female: {female_disease:.1f}% | Male: {male_disease:.1f}%
5. Data Quality: {missing_values} missing values, {duplicates} duplicates
6. Blood Pressure: Clear difference between disease and no-disease groups
7. Cholesterol: Higher levels show increased disease rate
8. Lifestyle: Physical activity shows protective effect
""")
```
