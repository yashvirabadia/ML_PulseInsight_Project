### -> IMPORT LIBRARIES

````python
import os
import shutil
import pickle
import pandas as pd
import numpy as np
from datetime import datetime

print("Libraries imported successfully!")
print(f"Current Date: {datetime.now().strftime('%Y-%m-%d %H:%M:%S')}")
````

### -> TEST MODEL LOADING

````python
# Test if model and scaler can be loaded
try:
    with open('../streamlit_app/best_model.pkl', 'rb') as f:
        model = pickle.load(f)
    print("Model loaded successfully!")
    print(f"Model Type: {type(model).__name__}")
    
    with open('../streamlit_app/scaler.pkl', 'rb') as f:
        scaler = pickle.load(f)
    print("Scaler loaded successfully!")
    print(f"Scaler Type: {type(scaler).__name__}")
    
except Exception as e:
    print(f" Error loading model/scaler: {e}")
````

### -> TEST PREDICTION FUNCTION

````python
# Test prediction with sample data
def test_prediction():
    # Sample patient data
    age_days = 50 * 365  # 50 years
    gender = 1  # Male
    height = 170  # cm
    weight = 75  # kg
    ap_hi = 130
    ap_lo = 85
    cholesterol = 2
    gluc = 1
    smoke = 0
    alco = 0
    active = 1
    
    # Calculate derived features
    bmi = weight / ((height / 100) ** 2)
    pulse_pressure = ap_hi - ap_lo
    map_value = ap_lo + (pulse_pressure / 3)
    
    # Create feature array
    features = np.array([[age_days, gender, height, weight, ap_hi, ap_lo,
                         cholesterol, gluc, smoke, alco, active,
                         bmi, pulse_pressure, map_value]])
    
    # Scale and predict
    features_scaled = scaler.transform(features)
    prediction = model.predict(features_scaled)[0]
    probability = model.predict_proba(features_scaled)[0]
    
    return prediction, probability

# Run test
try:
    pred, prob = test_prediction()
    print("Prediction Test Successful!")
    print(f"Prediction: {pred} ({'Disease' if pred == 1 else 'No Disease'})")
    print(f"Probability: No Disease={prob[0]:.2%}, Disease={prob[1]:.2%}")
except Exception as e:
    print(f"Prediction Test Failed: {e}")
````
